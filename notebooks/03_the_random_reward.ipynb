# The random reward control

Control C1 pays a reward drawn at random with the honest arm's mean, so it carries no information about the answer. Whatever it moves is what GRPO does with noise: the third hypothesis registered that it reaches less than a third of the honest gain on dev. This notebook draws both arms' confirmed accuracy at every checkpoint from the same start.

```python
import json
from pathlib import Path

import matplotlib

matplotlib.use("Agg")
import matplotlib.pyplot as plt

from goodhart_evaluation import arms

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
runs = arms.arm_runs(ROOT)
print("promoted defect study runs:", runs or "none yet")
```

```python
for arm in ("honest", "random"):
    run = runs.get(arm)
    if run is None:
        print(f"the {arm} arm has not run yet")
        continue
    dev = arms.earned(ROOT, run)
    if dev:
        start, end = dev[0][1], dev[-1][1]
        print(f"{arm}: twin confirmed on dev from {start:.3f} at step {dev[0][0]} to {end:.3f} at step {dev[-1][0]}")
    else:
        print(f"{arm}: no graded dev checkpoint yet")
```
